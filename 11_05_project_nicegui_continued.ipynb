```python
from nicegui import ui

def calculate():
    result = float(bill_amt.value)+float(bill_amt.value)*float(tip_pct.value)/100
    total_label.set_text(f"Total: ${result:.2f}")


bill_amt = ui.input("Bill amount")
tip_pct = ui.input("Tip %")
ui.button("Calculate",on_click=calculate)
total_label = ui.label("Total:")

ui.run(reload=False)
```

### Something new you should notice

We pass the name of the function `calculate` to the `on_click` parameter of the `ui.button`. This is a special feature of some programming languages called [first-class functions](https://en.wikipedia.org/wiki/First-class_function), where names of functions can be treated like other data. This is a hallmark of a style of programming called [functional programming](https://en.wikipedia.org/wiki/Functional_programming).

## Lambda Functions

Notice that in the NiceGUI documentation, you see examples that use `lambda` functions like this:

```python
from nicegui import ui

ui.button('Click me!', on_click=lambda: ui.notify('You clicked me!'))

ui.run(reload=False)
```

A **lambda function** is a function that doesn't have a name. You can write a small function inline without a `def` statement and it can be passed as an argument.

This has the same effect as if we had written

```python
from nicegui import ui

def click_notify():
    return ui.notify('You clicked me!')

ui.button('Click me!', on_click=click_notify)

ui.run(reload=False)
```
