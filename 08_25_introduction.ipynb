## Artificial Intelligence (AI) Policy

AI Large Language Models are capable of generating programming source code from natural language instructions, and they are an increasingly prevalent and important tool in modern software development. In order to make the best use of these tools, engineers first need to develop a proficiency in foundational computer science problem-solving principles. Early reliance on AI code generation inhibits this development. Thus, for code you turn-in for credit in this class, **you may not include AI-generated code** and you **may not use any code that was modified, fixed, or debugged by AI**.


### Suggestions on using and not using AI

If you choose to use AI as a learning aid in this course, you need to proceed with caution in order to ensure you're not falling into a violation of the course AI policy. You should never just paste the assignment into an AI chat application. I suggest always include the instruction "Do not generate any code" in your prompt. Here are some examples of prompts that *could be* acceptable uses of AI:

* "Ask me questions that will help me figure out why this code isn't doing what I want. Do not generate any code in your response."
* "Explain what this error message means, but do not fix my program and do not generate any code."
* "Here is what I think this code is doing: _____ . Is my explanation correct? Do not generate any code."

Because many Internet search engines include AI overviews/summaries/answers in their results, when searching for information about programming related to the class, you should use a search engine with AI features disabled like https://noai.duckduckgo.com/

### Editors

Note that many code editors have AI features built-in and on by default, including in ways that may not be apparent such as AI-powered autocomplete. Because of this, you are not permitted to use any coding environments other than Thonny, IDLE, the editor on the online grading platform, or another program that I have explicitly approved. If there is another program you want to use for your work, please come see me to discuss whether it is appropriate.
