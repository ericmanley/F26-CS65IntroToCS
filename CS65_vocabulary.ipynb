## CS 65: Introduction to Computer Science I

## Studying for November 17th Quiz

The quiz will check your familiarity with these terms. *I am not expecting you to memorize a formal definition for these.*

Here are some examples of kinds of questions you could expect along with possible good answers

**Q:** What is the name of this kind of statement: `my_var = 10`\
**A:** an assignment statement

**Q:** When you access an item in a dictionary like `phonebook["Eric"]`, what is the name for the thing you put inside the `[ ]`?\
**A:** key

**Q:** When you indent several lines of code under an `if`, a loop, or a function; what is the word for these lines of code?\
**A:** A block

**Q:** What is an accumulator?\
**A:** a variable that you use to keep a running total of a bunch of things

**Q:** What does it mean if a piece of data is immutable?\
**A:** you can't change it

**Q:** What is a local variable?\
**A:** a variable inside of a function that can only be used in that function
