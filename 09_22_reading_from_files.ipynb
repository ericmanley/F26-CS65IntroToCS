## Reading data from files

Python files work with several different methods that allow you to read data.
* `read()` reads the file into a big string
* `readline()` reads the next line of the file into a string
* `readlines()` reads the lines of the file into a list of strings

```python
gettysburg_file = open("gettysburg.txt")
file_contents = gettysburg_file.read()
print(file_contents)
```

```python
gettysburg_file = open("gettysburg.txt")
firstline = gettysburg_file.readline()
secondline = gettysburg_file.readline()
print(secondline)
```

```python
gettysburg_file = open("gettysburg.txt")
contents_as_list = gettysburg_file.readlines()
print( contents_as_list[8] )
```

## Opening files using `with` statements

Opening files with a `with` statement does the same thing, but it does some nice things like closing the file when it is done.

```python
with open("gettysburg.txt") as gettysburg_file:
    
    gettysburg_text = gettysburg_file.readlines()
    
    print(gettysburg_text)
    
print("All done with the file.")
```

## The newline character `\n`

Notice all of the lines from the file end in `\n`. What is that?

When the backslash character `\` is used inside of a string, it's a placeholder (called an **escape character**) for some special character
* `\n` is the newline character
* `\t` is the tab character
* `\"` allows you to put a quote character in the middle of a string
* `\\` is for putting an actual backslash in a string

So the `\n` is stored in the file so that file editors know where the author hit enter/return to go to the next line.

## Removing all the newlines

Suppose we wanted to remove the newline from every line in this list. We might plan out how to do this using *pseudocode* - a code-like notation that allows us to write down the idea behind an algorithm without getting bogged down in specific syntax.

1. open the file
2. read the lines of the file into a list variable
3. looping through the list, do the following with each item
    * strip out the newline character from the string
    * save the stripped version of the string back to the same spot in the list

Writing it in actual code might then look like this

```python
with open("top_boy_baby_names_2010s.txt") as boy_names_file:
    
    boy_names = boy_names_file.readlines()
    
    name_counter = 0
    
    while name_counter < len(boy_names):
        boy_names[name_counter] = boy_names[name_counter].rstrip()
        
        name_counter += 1
     
 
    print(boy_names)
```

## Now we can use Python to ask interesting question of our data

Here's a program that lets the user check how popular a name was.

```python
name_to_search = input("Enter a name: ")
if name_to_search in boy_names:
    position = boy_names.index(name_to_search)
    print(name_to_search,"was the number",(position+1),"most popular boy name in the 2010s.")
else:
    print(name_to_search,"was not a popular name in the 2010s.")
```
