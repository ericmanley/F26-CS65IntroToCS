## Formatting output with F-strings

You may have noticed that you sometimes get output that looks like this:

<p>
<div>
<img src="images/paycalcsample2.png" width="500"/>
</div>
</p>

There are two things here that aren't very user friendly:
1. We always end up with a space after a the `$`
2. It might show more decimal places than we want.


Fortunately, Python provides us with a way to control the format of strings using *f-strings*. F-strings start with an `f` before the opening `"`, and they let you embed the values of variables using `{}` like this.



```python
num_students = 42
print(f"We have {num_students} students going on the field trip.")
```

It also let's you specify some extra fomatting options for a variable by using a `:` and a special string formatter like `.4f` which means, *show 4 digits to the right of the decimal point in this float*. 

```python
pi_many_digits = 3.141592653589793238462643383279
print(f"pi rounded to 4 decimal places is {pi_many_digits:.4f}")
```

See here for more examples of string formatting that you can do: https://www.w3schools.com/python/python_string_formatting.asp
