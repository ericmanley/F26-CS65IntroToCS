## Graphical User Interfaces in Python

A __Graphical User Interface (GUI)__ is a means for a user to interact with a program which uses graphical components like windows, buttons, and other components that are manipulated with a mouse or touch-input device.

There are several different Python packages that provide the ability to build GUIs (some are designed for desktop/laptop apps, mobile, web, etc.)

**NiceGUI** is a framework that allows us to write Python web applications

## Minimal App

Import the `ui` *user interface* object from the `nicegui` module

Create a *label* object in the user interface with some text
* A **label** is a user interface component used to display text to the user

Run the app with `ui.run()`. We add the `reload=False` optional parameter to make sure it runs ok in Thonny. 

```python
from nicegui import ui

greeting_label = ui.label("Hello from NiceGUI!")

ui.run(reload=False)
```

This should open a web browser running your app. Click the Thonny **stop** button to stop it. 



## A slightly larger example



```python
from nicegui import ui

def greet():
    result_label.set_text(f"{name_input.value} clicked the button")


name_input = ui.input("Name")
ui.button("Click me!", on_click=greet)
result_label = ui.label("Your text will appear here")

ui.run(reload=False)
```

## Callback Functions

A **callback** function is a function that runs in response to some user interface event (like a button click).

Here `greet()` is a callback function and the `on_click=greet` (no parantheses!) is how we tell the button which function to run when it is clicked.
