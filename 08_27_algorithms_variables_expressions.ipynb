## Programs and Programming Languages

__Program__: a computer program is a series of instructions for the computer to execute.

An algorithm is an _idea_ and the program is a concrete _expression_ of that idea.

__Programming Language__: a set of rules that define which programs are legal

Like natural languages, programming languages have a _grammar_ and _syntax_ - they describe how instructions are constructed

## Rules for variable names

__Allowed:__ upper and lowercase letters, numbers, underscores _

__Not Allowed:__ other special characters, spaces

```python
total$ = 42.0
```

Can't _start_ with a number

```python
5th_grade_students = 42
```

Names a _case-sensitive_ (lower/uppercase matters)

```python
students = 42
print(Students) 
```

## Assignment Statements

Code with a `=` is called an __assignment statement__

```python
students = 42
```

Whatever is on the _right_ side of `=` will be evaluated and saved to the variable named on the _left_ side.

You can't write it the other way around.

```python
42 = students
```
