## Lists as arguments and return values

You can normally use lists as arguments and return values just fine, though there's a slight nuance to be aware of. 

````python
def double_val(param):
    param = param * 2

x = 4
double_val(x)
print(x)
````

````python
def double_list(param):
    param[0] = param[0] * 2

x = [4]
double_list(x)
print(x)
````

## Why does this happen?

When you pass an argument to a parameter, it creates an **alias** - another name bound to the same thing in the computer's memory.

There's a difference between assigning a new value to the alias variable (making it no longer an alias)

vs.

using the alias to change someting in the original location in memory.
