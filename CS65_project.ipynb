## Finding Data

There are many sources for good data. If you interested in a particular topic like a sport, a game, music, food, etc., you can usually just search for keywords associated with your interest and the terms *csv file* or *json file*. You can also use Google's dataset search feature: https://datasetsearch.research.google.com/

Here are some other good sources of datasets:
* **CORGIS Datasets Project:** https://corgis-edu.github.io/corgis/ . This is a collection of curated datasets that come in both csv and json formats meant to be used with programming projects
* **TidyTuesday:** https://github.com/rfordatascience/tidytuesday . This is a project for practicing data science using real, interesting data sets that are posted each week. It's mean for R programmers, which is another language like Python, but the data is usually posted in csv format
* **FiveThirtyEight:** https://github.com/fivethirtyeight/data . This is data used in news articles on topics like sports, politics, and pop culture. It's usually available in csv format.
* **Kaggle:** https://www.kaggle.com/datasets . This is a site that hosts data science competitions and has a user community for sharing all kinds of data. Data quality varies for community contributions.
* **Iowa Data Hub:** https://data.iowa.gov/ . This is the state of Iowa's data portal that provides access to various governmental data sets. Many of them are downloadable in csv and/or json formats.

## What should my program do with the data?

If you're having trouble coming up with an idea for your program, think about things that users might like to do with the data. For example
* search or filter the data for something of interest
* compare or visualize data items
* get a random pick
* take a quiz
* compute a total or find the largest/smallest value
